```python
import random
import matplotlib.pyplot as plt
```

```python
def abordar_avion(numero_asientos=100):
    asientos_disponibles = list(range(1, numero_asientos + 1))

    # Pasajero 1: elige completamente al azar
    asiento_elegido = random.choice(asientos_disponibles)
    asientos_disponibles.remove(asiento_elegido)

    asiento_ultimo_pasajero = None

    # Pasajeros 2, 3, ..., 100
    for pasajero in range(2, numero_asientos + 1):
        if pasajero in asientos_disponibles:
            asiento_elegido = pasajero
        else:
            asiento_elegido = random.choice(asientos_disponibles)

        asientos_disponibles.remove(asiento_elegido)

        if pasajero == numero_asientos:
            asiento_ultimo_pasajero = asiento_elegido

    return asiento_ultimo_pasajero
```

```python
numero_simulaciones = 100000
numero_asientos = 100

frecuencias = [0] * numero_asientos

for _ in range(numero_simulaciones):
    asiento_final = abordar_avion(numero_asientos)
    frecuencias[asiento_final - 1] += 1

veces_en_su_asiento = frecuencias[99]
probabilidad = veces_en_su_asiento / numero_simulaciones

print(f"Número de simulaciones: {numero_simulaciones}")
print(f"Veces que el pasajero 100 terminó en el asiento 100: {veces_en_su_asiento}")
print(f"Probabilidad estimada: {probabilidad:.4f}")
print(f"Probabilidad teórica: {0.5:.4f}")
```

```python
asientos = list(range(1, numero_asientos + 1))

plt.figure(figsize=(16, 6))
plt.bar(asientos, frecuencias)
plt.xlabel("Asiento en el que termina el pasajero 100")
plt.ylabel("Frecuencia")
plt.title("Distribución del asiento final del pasajero 100")
plt.xticks(range(1, 101, 5))
plt.grid(axis="y", alpha=0.3)
plt.show()
```
