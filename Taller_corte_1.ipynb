# 3. Realiza un algoritmo que transforme numeros romanos a numeros decimales

```python
numeros_romanos = {
    "I": 1,
    "V": 5,
    "X": 10,
    "L": 50,
    "C": 100,
    "D": 500,
    "M": 1000
}
numero = input("Ingrese un número romano: ")
try:
    for char in numero:
        if char.upper() not in numeros_romanos:
            raise ValueError(f"El carácter '{char}' no es un número romano válido.")
except ValueError as e:
    print(f"Error: {e} el valor ingresado no es un número romano.")


def romano_a_decimal(romano):
    romano = romano.upper()
    
    decimal = 0

    for i in range(len(romano)):
        valor_actual = numeros_romanos[romano[i]]

        if i + 1 < len(romano):
            valor_siguiente = numeros_romanos[romano[i + 1]]

            if valor_actual < valor_siguiente:
                decimal = decimal - valor_actual
            else:
                decimal = decimal + valor_actual
        else:
            decimal = decimal + valor_actual

    return decimal



resultado = romano_a_decimal(numero)

print(f"{numero.upper()} en decimal es: {resultado}")
```

# 5. Realiza un algoritmo que valide si una frase es un palindromo

```python
input_string = input("Ingrese una palabra: ")

try:
    if not input_string.strip().isalpha():
        raise ValueError("La palabra debe contener solo letras.")

except ValueError as e:
    print(f"Error: {e} El valor ingresado no es una palabra válida.")

else:
    palabra = input_string.strip().lower().replace(" ", "")

    if palabra == palabra[::-1]:
        print(f"{palabra} es un palíndromo.")
    else:
        print(f"{palabra} no es un palíndromo.")
```

# 6. Realiza un algoritmo que cuente la cantidad de ocurrencias de cada caracter en una frase

```python
frase = input("Ingrese una frase: ").lower()

ocurrencias = {}

for caracter in frase:

    if caracter != " ":

        if caracter in ocurrencias:
            ocurrencias[caracter] = ocurrencias[caracter] + 1
        else:
            ocurrencias[caracter] = 1

print(ocurrencias)
```
